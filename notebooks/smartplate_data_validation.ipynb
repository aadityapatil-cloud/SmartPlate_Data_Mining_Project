```python
# ============================================================
# STEP 7: CHECK MISSING VALUES
# ============================================================

# isnull() identifies missing values
# sum() counts missing values in each column

missing_values = df.isnull().sum()

print(missing_values)

# Calculate total number of missing values
total_missing = missing_values.sum()

print("\nTotal Missing Values:", total_missing)
```

```python
# ============================================================
# STEP 8: VALIDATE MISSING VALUES
# ============================================================

# If there are no missing values, validation passes
if total_missing == 0:

    print("✅ PASS: No missing values.")

else:

    print("⚠️ WARNING: Missing values found.")
```

```python
# ============================================================
# STEP 9: CHECK DUPLICATE ROWS
# ============================================================

# duplicated() identifies duplicate records
# sum() counts the total number of duplicate rows

duplicate_rows = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_rows)
```

```python
# ============================================================
# STEP 10: VALIDATE DUPLICATE ROWS
# ============================================================

# Check whether the dataset contains duplicate records

if duplicate_rows == 0:

    print("✅ PASS: No duplicate rows found.")

else:

    print("⚠️ WARNING: Duplicate rows found.")
```
