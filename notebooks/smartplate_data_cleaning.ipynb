```python
# ============================================================
# STEP 1: IMPORT LIBRARY
# ============================================================

import pandas as pd
import numpy as np
```

```python
# ============================================================
# STEP 2: DATASET PATH
# ============================================================

INPUT_FILE = r"C:\Users\Aaditya babaso patil\Desktop\SmartPlate_Data_Mining_Project\data\smartplate_cafeteria_dataset.csv"
```

```python
# ============================================================
# STEP 3: LOAD DATASET
# ============================================================

df = pd.read_csv(INPUT_FILE)

print(" Dataset loaded successfully.")
print("Original Shape:", df.shape)
```

```python
# ============================================================
# STEP 4: CHECK MISSING VALUES
# ============================================================

# Count missing values in each column
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values)
```

```python
# ============================================================
# STEP 5: HANDLE MISSING VALUES
# ============================================================

# Separate numeric and categorical columns
numeric_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

# Fill numeric missing values with the median
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

# Fill categorical missing values with the mode
for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print(" Missing values handled.")
```

```python
# ============================================================
# STEP 6: REMOVE DUPLICATE ROWS
# ============================================================

# Count duplicates before removing them
duplicate_count = df.duplicated().sum()

print("Duplicate Rows Before Cleaning:", duplicate_count)

# Remove duplicate rows
df = df.drop_duplicates()

print("Duplicate Rows After Cleaning:", df.duplicated().sum())
```

```python
# ============================================================
# STEP 7: REMOVE EXTRA SPACES FROM TEXT
# ============================================================

# Clean spaces from all text columns
for column in categorical_columns:
    df[column] = df[column].astype(str).str.strip()

print(" Extra spaces removed.")
```

```python
# ============================================================
# STEP 8: STANDARDIZE TEXT VALUES
# ============================================================

# Convert text columns to a consistent format
for column in categorical_columns:
    df[column] = df[column].str.title()

print(" Text values standardized.")
```

```python
# ============================================================
# STEP 9: CHECK NEGATIVE VALUES
# ============================================================

# Find negative values in numeric columns
for column in numeric_columns:

    negative_count = (df[column] < 0).sum()

    print(
        f"{column}: "
        f"{negative_count} negative values"
    )
```

```python
# ============================================================
# STEP 10: HANDLE NEGATIVE VALUES
# ============================================================

# Replace negative numeric values with NaN
for column in numeric_columns:
    df.loc[df[column] < 0, column] = np.nan

# Fill newly created missing values with median
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Negative values handled.")
```

```python
# ============================================================
# STEP 11: CONVERT DATE COLUMN
# ============================================================

# Check whether a Date column exists
if "Date" in df.columns:

    df["Date"] = pd.to_datetime(
        df["Date"],
        errors="coerce"
    )

    print("Date column converted to datetime.")

else:

    print(" Date column not found.")
```

```python
# ============================================================
# STEP 12: FINAL MISSING VALUE CHECK
# ============================================================

print("Final Missing Values:")
print(df.isnull().sum())

print(
    "\nTotal Missing Values:",
    df.isnull().sum().sum()
)
```

```python
# ============================================================
# STEP 13: FINAL DUPLICATE CHECK
# ============================================================

print(
    "Final Duplicate Rows:",
    df.duplicated().sum()
)
```

```python
# ============================================================
# STEP 14: CLEANED DATASET
# ============================================================

print("Cleaned Dataset Shape:", df.shape)

df.head()
```

```python
# ============================================================
# STEP 15: SAVE CLEANED DATASET
# ============================================================

OUTPUT_FILE = r"C:\Users\Aaditya babaso patil\Desktop\SmartPlate_Data_Mining_Project\data\smartplate_cafeteria_cleaned.csv"

df.to_csv(
    OUTPUT_FILE,
    index=False
)

print(" Cleaned dataset saved successfully.")
print("File:", OUTPUT_FILE)
```
